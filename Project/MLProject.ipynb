## model building

```python
x = df.drop('is_claim',axis=1)
y =df['is_claim']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegressionCV
 
def logistic_model():
     model = LogisticRegressionCV()
     model.fit(x_train,y_train)
     return model
```

```python
from sklearn.neighbors import KNeighborsClassifier

def knn_model():
    model = KNeighborsClassifier(n_neighbors=7)
    model.fit(x_train,y_train)
    return model
```

```python
from sklearn.svm import SVC

def svm_model():
    model = SVC()
    model.fit(x_train,y_train)
    return model
```

```python
from sklearn.tree import DecisionTreeClassifier

def dtree_model():
    model = DecisionTreeClassifier()
    model.fit(x_train,y_train)
    return model
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted')
    recall = recall_score(y_test, y_pred, average='weighted')
    f1 = f1_score(y_test, y_pred, average='weighted')

    return accuracy, precision, recall, f1
```

```python
models = []

models.append({"name": "Logistic Regression", "model": logistic_model()})
models.append({"name": "KNN", "model": knn_model()})
models.append({"name": "SVM", "model": svm_model()})
models.append({"name": "Decision Tree", "model": dtree_model()})
```

```python
for info in models:
    accuracy, precision, recall, f1 = evaluate_model(
        info["model"], x_test, y_test
    )
    
    info["accuracy"] = accuracy
    info["precision"] = precision
    info["recall"] = recall
    info["f1"] = f1


```

```python
# print the evaluation report
df_evaluation_report = pd.DataFrame(models, columns=["name", "accuracy", "precision", "recall", "f1"])
df_evaluation_report
```
