#### 2. Write a NumPy program to create an element-wise comparison (greater,    greater_equal, less and less_equal) of two given arrays.

```python
import numpy as np
a1 = np.array([1, 2, 3, 4])
a2 = np.array([5, 6, 7, 8])
print(a1[0] > a2[0])
print(a1[3] >= a2[2])
print(a1[2] < a1[1])
print(a1[1] <= a2[2]) 
```

#### 6. Write a NumPy program to get the element-wise remainder of an array of numbers from 20 to 50 which are divisible by 3.

```python
import numpy as np

# create array from 20 to 50
arr = np.arange(20, 51)

# select numbers divisible by 3
div_by_3 = arr[arr % 3 == 0]

# element-wise remainder (mod 3)
remainder = div_by_3 % 3

print("Numbers divisible by 3:", div_by_3)
print("Element-wise remainder:", remainder)
```

#### 7. Write a Pandas program to add, subtract, multiple and divide two Pandas Series.
   Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]


```python
import numpy as np
import pandas as pd

s1 = pd.Series([2, 4, 6, 8, 10])
s2 = pd.Series([1, 3, 5, 7, 9])

print(f"Add: \n{s1 + s2}\n")
print(f"Subtract: \n{s1 - s2}\n")
print(f"Multiply: \n{s1 * s2}\n")
print(f"Division: \n{s1 / s2}\n")
```

```python
# 11
df = pd.read_csv("mtcars.csv")
df
```

```python
# i)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.plot(df["wt"], df["mpg"], label = "Milage", color = "red")
plt.title("Weight vs Milage")
plt.xlabel("Weight")
plt.ylabel("Milage")
plt.grid(alpha = 0.4)

plt.legend()
plt.show()
```

```python
# ii)
avg_mpg = df.groupby("cyl")["mpg"].mean()

plt.bar(avg_mpg.index, avg_mpg.values)
plt.title("Bar Graph: Average MPG by Cylinders")
plt.xlabel("Number of Cylinders")
plt.ylabel("Average MPG")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# iii)
cyl_counts = df["cyl"].value_counts()

plt.pie(
    cyl_counts.values,
    labels=cyl_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Pie Chart: Cylinder Distribution")
plt.show()
```

```python
# iv)
plt.boxplot(df["mpg"])
plt.title("Boxplot: Mileage (mpg)")
plt.ylabel("MPG")
plt.grid(alpha=0.3)
plt.show()
```

```python
# v)
sns.pairplot(df)
plt.show()
```

```python
# vi)
plt.scatter(df["hp"], df["mpg"])
plt.title("Scatter Plot: Horsepower vs Mileage")
plt.xlabel("Horsepower (hp)")
plt.ylabel("Mileage (mpg)")
plt.grid(alpha=0.4)
plt.show()
```

```python
# vii)
plt.hist(df["mpg"], bins=10)
plt.title("Histogram: Mileage Distribution")
plt.xlabel("MPG")
plt.ylabel("Frequency")
plt.grid(alpha=0.3)
plt.show()
```
