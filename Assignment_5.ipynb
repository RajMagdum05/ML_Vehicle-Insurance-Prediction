```python
# In this dataset you have to predict that customer is exited or not based on various factors given in dataset.

# Dataset : Churn.csv
# Steps to follow:
# 1. Do Proper EDA.
# 2. Drop Unnecessary Columns.
# 3. convert Categorical columns into numerical columns.

# Prepare model by using KNN & Calculate Accuracy by confusion matrix, Calculate accuracy score, precision score, recall score, f1 score.
```

```python
# classification problem using knn
```

```python
# import
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# read_csv
df = pd.read_csv("Datasets/Churn.csv")
```

```python
# EDA steps
df.shape
```

```python
df.info()
```

```python
df.head()
```

```python
df.tail()
```

```python
# checking missing values
df.isna().sum()
```

```python
# drop garbage columns
df.drop(["RowNumber", "CustomerId", "Surname"], axis = 1, inplace = True)
```

```python
df.head()
```

```python
# Understand each column
# Geography -> Nominal OneHot Encoding
# Gender -> Binary (0/1)
# others are numerical needs StandardScaler

numeric_cols = ["CreditScore", "Age", "Tenure", "Balance", "NumOfProducts", "HasCrCard", "IsActiveMember", "EstimatedSalary"]
nominal_cols = ["Geography"]
binary_cols = ["Gender"]
```

```python
# split data
x = df.drop("Exited", axis = 1)
y = df["Exited"]
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# columnTransformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers = [
        ("num", StandardScaler(), numeric_cols),
        ("cat", OneHotEncoder(handle_unknown = "ignore"), nominal_cols),
        ("bin", OneHotEncoder(drop='if_binary'), binary_cols)
    ],
    remainder = "drop"
)
```

```python
# pipeline
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier

pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", KNeighborsClassifier())
])
```

```python
# train model
pipeline.fit(x_train, y_train)
```

```python
# test and pred
y_pred = pipeline.predict(x_test)
y_pred
```

```python
# Evaluation / Accuracy
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score

cm = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Confusion Matrix = {cm}")
print(f"Accuracy Score = {accuracy * 100:.2f} %")
print(f"Precision Score = {precision * 100:.2f} %")
print(f"Recall Score = {recall * 100:.2f} %")
print(f"f1 Score = {f1 * 100:.2f} %")
```
