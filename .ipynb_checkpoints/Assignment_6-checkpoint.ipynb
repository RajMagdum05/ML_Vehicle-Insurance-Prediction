#### Q.1

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Que.1 -
# In this dataset you have to predict that customer is exited or not based on various factors given in dataset.

# Dataset : Churn.csv
# Steps to follow:
# 1. Do Proper EDA.
# 2. Drop Unnecessary Columns.
# 3. convert Categorical columns into numerical columns.

# Prepare model by using SVM & Decision Tree & Calculate Accuracy by confusion matrix, Calculate accuracy score, precision score, recall score, f1 score.
```

```python
# read csv
df = pd.read_csv("Datasets/Churn.csv")
```

```python
# basic EDA steps
df.shape
```

```python
df.info()
```

```python
df.head()
```

```python
# check NaN values
df.isna().sum()
```

```python
# no NaN values 
```

```python
# dropping useless/garbage columns
df.drop(["RowNumber", "Surname", "CustomerId"], axis = 1, inplace = True)
```

```python
df.head()
```

```python
# Understanding each column
# nominal - Geography
# binary - Gender
# remaining all numeric
```

```python
# Defining columns
numeric_cols = (df.drop(["Geography", "Gender", "Exited"], axis = 1)).select_dtypes(include = ["int64", "float64"]).columns.tolist()
nominal_cols = ["Geography"]
binary_cols = ["Gender"]
```

```python
# split data
x = df.drop("Exited", axis = 1)
y = df["Exited"]
```

```python
# train_test_split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# ColumnTransformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers = [
        ("num", StandardScaler(), numeric_cols),
        ("cat", OneHotEncoder(handle_unknown = "ignore"), nominal_cols),
        ("bin", OneHotEncoder(drop = "if_binary"), binary_cols)
    ],
    remainder = "drop"
)
```

```python
# pipeline
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC

pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", SVC())
])
```

```python
# train model
pipeline.fit(x_train, y_train)
```

```python
# test model
y_pred = pipeline.predict(x_test)
y_pred
```

```python
# Evaluation
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
cm = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Confusion Matrix = {cm}")
print(f"Accuracy Score = {accuracy * 100:.2f} %")
print(f"Precision Score = {precision * 100:.2f} %")
print(f"Recall Score = {recall * 100:.2f} %")
print(f"f1 Score = {f1 * 100:.2f} %")
```

#### Q.2

```python
# Que2.-
# In this dataset you have to predict that whether we have to give ATTORNEY to the clients or not (0=Yes, 1=No) based on below factors.
# CLMAGE - Age of Client
# CLMSEX - Sex of client (0=Male, 1= Female)
# SEATBELT - Whether client has Seatbelt or not (0=Yes, 1=No)
# CLMINSUR - Whether Client give Insurance or not (0=Yes, 1=No)
# LOSS - Loss that client faced

# Prepare model by using SVM & Decision Tree & Calculate Accuracy by confusion matrix, Calculate accuracy score, precision score, recall score, f1 score.

# Dataset : Claimants.csv
```

```python
# read csv
df = pd.read_csv("Datasets/claimants.csv")
```

```python
# basic EDA steps
df.shape
```

```python
df.info()
```

```python
df.head()
```

```python
# check NaN values
df.isna().sum()
```

```python
# dropna
df.dropna(axis = 0, inplace = True) # do inplace = True here
```

```python
df.isna().sum()
```

```python
# no garbage/useless columns in this dataset
```

```python
# understanding columns
# output column - ATTORNEY
# all are numeric columns - StandardScaler works well
numeric_cols = (df.drop("ATTORNEY", axis = 1)).select_dtypes(include = ["int64", "float64"]).columns.tolist()
```

```python
# split the data
x = df.drop("ATTORNEY", axis = 1)
y = df["ATTORNEY"]
```

```python
# train_test_split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# ColumnTransformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler

preprocessor = ColumnTransformer(
    transformers = [
        ("num", StandardScaler(), numeric_cols)
    ],
    remainder = "drop"
)
```

```python
# pipeline
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC

pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("model", SVC())
])
```

```python
# train model
pipeline.fit(x_train, y_train)
```

```python
# test / predict
y_pred = pipeline.predict(x_test)
y_pred
```

```python
#Evaluation
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
cm = confusion_matrix(y_test, y_pred)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Confusion Matrix = {cm}")
print(f"Accuracy Score = {accuracy * 100:.2f} %")
print(f"Precision Score = {precision * 100:.2f} %")
print(f"Recall Score = {recall * 100:.2f} %")
```
