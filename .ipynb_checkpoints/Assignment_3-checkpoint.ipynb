#### Que. 2
#### In the given dataset TVMarketing.csv you have given  sales based on 'TV' marketing budget.
#### Build a linear regression model to predict 'Sales' using 'TV' as the predictor variable. Also plot regression line.

#### Dataset : TVMarketing.csv

```python
df = pd.read_csv("Datasets/TvMarketing.csv")
```

```python
df.info()
df.describe()
df.columns
df.head()
```

```python
# output -> Sales
# input -> TV
```

```python
df.isna().sum()
```

```python
df.corr()
```

```python
# x and y
x = df.drop(["Sales"], axis=1)
y = df["TV"]
x.head()
y.head()
```

```python
# train_test_split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# model application
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
plt.scatter(x_test, y_test, color='blue', label='Actual data')
plt.plot(x_test, y_pred, color='red', label='Regression line')

plt.title("TV vs Sales")
plt.xlabel("TV")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

##### Que. 3
##### You own an ice cream business and you would like to create a model 
##### that could predict the daily revenue in dollars based on the outside air temperature (degC). 
##### Build Linear Regression Model and plot regression line.

##### Dataset : IceCreamData.csv

```python
# read csv
df = pd.read_csv("Datasets/IceCreamData.csv")

# EDA basic steps
df.info()
df.describe()
df.columns
df.head()
```

```python
# output - Revenue
# input - Temperature
```

```python
# NaN Value
df.isna().sum()
```

```python
# correlation
df.corr()
```

```python
# x and y
x = df.drop(["Revenue"], axis=1)
y = df["Revenue"]
x.head()
y.head()
```

```python
# train_test_split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
# applying model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
y_pred
```

```python
# plotting regression line
plt.scatter(x_test, y_test, label = "Regression Line", color = "blue")
plt.plot(x_test, y_pred, label = "Linear Regression", color = "red")

plt.title("Regression model Temp vs Revenue")
plt.xlabel("Temperature")
plt.ylabel("Revenue")
plt.legend
plt.show()
```
