```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings(action="ignore")
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score,f1_score,recall_score,precision_score,confusion_matrix
```

```python
df=pd.read_csv("C:\\Users\\yoges\\OneDrive\\Desktop\\sunbeam\\IIT-08-A-DS\\Datasets\\Churn.csv")
df
```

```python
df.head()
```

```python
df.tail()
```

```python
df.describe()
```

```python
df.info
```

```python
df.isna().sum()
```

```python
df.drop(['RowNumber','CustomerId','Surname'],axis=1,inplace=True)
```

```python
df.columns
```

```python
lb=LabelEncoder()
df['Geography']=lb.fit_transform(df['Geography'])
df['Gender']=lb.fit_transform(df['Gender'])
df
```

```python
x=df.drop('Exited',axis=1)
y=df['Exited']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8)
x_test
```

```python
model=DecisionTreeClassifier()
model.fit(x_train,y_train)
```

```python
parametres={'criterion':["gini","entropy","log_loss"],'max_depth':np.arange(10)*100}
grid_search_cv=GridSearchCV(estimator=model,param_grid=parametres)
grid_search_cv.fit(x_train,y_train)
```

```python
grid_search_cv.best_params_
```

```python
grid_search_cv.best_score_
```

```python
y_pred=grid_search_cv.predict(x_test)
y_pred
```

```python
cfg=confusion_matrix(y_test,y_pred)
cfg
```

```python
pre=precision_score(y_test,y_pred)
pre
```

```python
recall=recall_score(y_test,y_pred)
recall
```

```python
f1=f1_score(y_test,y_pred)
f1
```

```python
acc=accuracy_score(y_test,y_pred)
acc
```

```python
from sklearn.svm import SVC
```

```python
model=SVC()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```
